```python
class student:
    def __init__(self,name,age) -> None: 
        self.name = name 
        self.age = age
    @staticmethod
    def hello():
        print("hi ")
    def info(self):
        print("name is",self.name,"and age is",self.age)

s1 = student("raman","18") 
print(s1.info())
```

```python
class bank:
    def __init__(self,number,psw,name) -> None:
        self.number = number
        self.__psw = psw # private which cant be accessed # can also be done with func and attributes 
        self.name = name
    def reset(self):
        print("your reset is reset to ",self.__psw) # but can be accessed outside class
        
p1 = bank(1122,"hello@123","Shaurya")
print(p1.reset())
```

```python
class car:
    @staticmethod
    def start():
        print("starting vroom vroom")
class toyota(car):
    def __init__(self,name) -> None:
        self.name = name
c1 = toyota("fortuner")
print(c1.start())
```

```python
class car:
    def __init__(self,type) -> None:
        self.type = type
    @staticmethod
    def start():
        print("starting vroom vroom")
    
class toyota(car):
    def __init__(self,name,type) -> None:
        super().__init__(type)
        self.name = name
        super().start()
c1 = toyota("fortuner","diesal")
print(c1.type)
```

```python
class person:
    name = "unknown"
    
    @classmethod
    def real_name(cls,name):
        cls.name = name 
p1 = person()
p1.real_name("rohit") # this changes name at both class and object 
print(person.name) # otherwise this would have printed unknown 
# for instance check 
```

```python
class stu:
    def __init__(self,phy,chem,math):
        self.phy = phy
        self.chem = chem
        self.math = math

    @property # here if we will latter change any sub marks this will take latest data 
    def perc(self):
        return str((self.phy + self.chem + self.math) / 3)

s1 = stu(99,98,11)
print(s1.perc)
#getter and setter decorator 
```

```python
class Complex:
    def __init__(self, real, img):
        self.real = real
        self.img = img

    def showNumber(self):
        print(self.real, "i +", self.img, "j")

    def __add__(self, num2):
        newReal = self.real + num2.real
        newImg = self.img + num2.img
        return Complex(newReal, newImg)

    def __sub__(self, num2):
        newReal = self.real - num2.real
        newImg = self.img - num2.img
        return Complex(newReal, newImg)


num1 = Complex(1, 3)
num1.showNumber()
num2 = Complex(2,3)
num3 = num1 +num2
print(num3.showNumber())
```
