```python
class student:
    marks = -1

    def __init__(self, name, *args, **kwargs):
        self.name = name
        self.marks = args
        self.info = kwargs

    def details(self):
        print("Name:", self.name)
        print("Marks:", self.marks)
        print("Additional info:", self.info)

    def calc_avg(self):
        if len(self.marks) == 0:
            return 0

        return sum(self.marks) / len(self.marks)

    def is_pass(self):
        if any(mark < 0 for mark in self.marks):
            return False

        if len(self.marks) == 0:
            return False

        return self.calc_avg() >= 40
s1 = student("Rahul", 80, 75, 90, age=21, city="Delhi")

s1.details()
print("Average:", s1.calc_avg())
print("Passed:", s1.is_pass())
```

```python
class Employee:
    def __init__(self, name, emp_id, salary):
        self.name = name
        self.emp_id = emp_id
        self.salary = salary

    def details(self):
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Salary:", self.salary)


class Developer(Employee):
    def __init__(self, name, emp_id, salary, *args, **kwargs):
        super().__init__(name, emp_id, salary)
        self.lang = args
        self.exp = kwargs

    def details(self):
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Salary:", self.salary)
        print("Languages:", self.lang)
        print("Additional info:", self.exp)


# Testing
emp1 = Employee("Rahul", 101, 50000)
emp1.details()

print()

dev1 = Developer(
    "Priya", 102, 70000,
    "Python", "C++",
    experience=2, team="AI"
)
dev1.details()
```
